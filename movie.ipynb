```python
# Reload data fresh to avoid column issues from re-running cells
movies = pd.read_csv('tmdb_5000_movies.csv')
credits = pd.read_csv('tmdb_5000_credits.csv')

# Merge
movies = movies.merge(credits, on='title')

# Select relevant columns
movies = movies[['movie_id','title','overview','genres','keywords','cast','crew']]

# Drop nulls and duplicates
movies.dropna(inplace=True)

# Convert genres and keywords
movies['genres'] = movies['genres'].apply(convert)
movies['keywords'] = movies['keywords'].apply(convert)

# Convert cast - get top 3 actors
def convert3(obj):
    L = []
    for i in ast.literal_eval(obj)[:3]:
        L.append(i['name'])
    return L

movies['cast'] = movies['cast'].apply(convert3)
movies.head()
```

```python
# Fetch the director from crew
def fetch_director(obj):
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            return [i['name']]
    return []

movies['crew'] = movies['crew'].apply(fetch_director)
movies.head()
```

```python
# Convert overview from string to list (handle if already converted)
movies['overview'] = movies['overview'].apply(lambda x: x.split() if isinstance(x, str) else x)
movies.head()
```

```python
# Remove spaces from multi-word names (e.g., "Sam Worthington" -> "SamWorthington")
# This prevents confusion during vectorization
movies['genres'] = movies['genres'].apply(lambda x: [i.replace(" ", "") for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x: [i.replace(" ", "") for i in x])
movies['cast'] = movies['cast'].apply(lambda x: [i.replace(" ", "") for i in x])
movies['crew'] = movies['crew'].apply(lambda x: [i.replace(" ", "") for i in x])
movies.head()
```

```python
# Create a 'tags' column by combining all features
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
movies.head()
```

```python
# Create new DataFrame with only needed columns
new_df = movies[['movie_id', 'title', 'tags']].copy()

# Convert tags list to a single string (lowercase)
new_df['tags'] = new_df['tags'].apply(lambda x: " ".join(x).lower())
new_df.head()
```

```python
# Apply stemming to reduce words to root form
from nltk.stem.porter import PorterStemmer
import nltk
nltk.download('punkt', quiet=True)

ps = PorterStemmer()

def stem(text):
    return " ".join([ps.stem(word) for word in text.split()])

new_df['tags'] = new_df['tags'].apply(stem)
new_df['tags'].iloc[0]
```

```python
# Vectorize tags using CountVectorizer (Bag of Words)
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

cv = CountVectorizer(max_features=5000, stop_words='english')
vectors = cv.fit_transform(new_df['tags']).toarray()

print(f"Vector shape: {vectors.shape}")
print(f"Feature names sample: {cv.get_feature_names_out()[:10]}")
```

```python
# Compute cosine similarity between all movies
similarity = cosine_similarity(vectors)
print(f"Similarity matrix shape: {similarity.shape}")
print(f"Similarity of movie 0 with movie 1: {similarity[0][1]:.4f}")
```

```python
# Recommendation function
def recommend(movie):
    movie_index = new_df[new_df['title'] == movie].index[0]
    distances = similarity[movie_index]
    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:6]
    
    print(f"\n🎬 Top 5 movies similar to '{movie}':\n")
    for i, (idx, score) in enumerate(movies_list, 1):
        print(f"  {i}. {new_df.iloc[idx]['title']} (Similarity: {score:.4f})")

# Test with a few movies
recommend('Batman Begins')
recommend('Avatar')
recommend('The Dark Knight Rises')
```

```python
# Save model data for Streamlit app
import pickle

pickle.dump(new_df.to_dict(), open('movies_dict.pkl', 'wb'))
pickle.dump(similarity, open('similarity.pkl', 'wb'))

print("✅ Model data saved successfully!")
print(f"   - movies_dict.pkl")
print(f"   - similarity.pkl")
print(f"   - Total movies: {len(new_df)}")
```
